# 02. EDA — Exploratory Data Analysis

```python
import sys
sys.path.insert(0, "../src")
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns, numpy as np
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("../data/raw/boxing_dataset.csv")
df.head()
```

```python
# Distribusi Target
counts = df["performance_level"].value_counts()[["Low","Medium","Peak"]]
colors = {"Low":"#e74c3c","Medium":"#f39c12","Peak":"#2ecc71"}
fig,ax=plt.subplots(figsize=(7,4))
ax.bar(counts.index, counts.values, color=[colors[k] for k in counts.index])
ax.set_title("Performance Level Distribution")
plt.show()
```

```python
# Tren mingguan
features=["cortisol_nmol_L","hrv_ms","vo2max_ml_kg_min","punch_speed_m_s","endurance_score","training_load"]
weekly=df.groupby("week")[features].mean()
fig,axes=plt.subplots(len(features),1,figsize=(12,len(features)*2.5),sharex=True)
for ax,feat in zip(axes,features):
    ax.plot(weekly.index,weekly[feat],marker="o",linewidth=2)
    ax.set_ylabel(feat,fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
numeric_cols=["cortisol_nmol_L","testosterone_nmol_L","ck_u_L","lactate_mmol_L","hrv_ms","vo2max_ml_kg_min","reaction_time_ms","punch_speed_m_s","endurance_score","strength_index"]
corr=df[numeric_cols].corr()
mask=np.triu(np.ones_like(corr,dtype=bool))
fig,ax=plt.subplots(figsize=(12,9))
sns.heatmap(corr,mask=mask,annot=True,fmt=".2f",cmap="RdYlGn",center=0,ax=ax)
plt.show()
```

```python
# Boxplot per faze
phase_order=["GPP","SPP","Competition","Taper"]
fig,ax=plt.subplots(figsize=(8,5))
sns.boxplot(data=df,x="phase",y="hrv_ms",order=phase_order,palette="Set2",ax=ax)
ax.set_title("HRV per Fase")
plt.show()
```
